```python
import torch

# target output
y_target = torch.tensor([[1.0, 0.5]])

# model layer
W = torch.randn(2, 2, requires_grad=False)
b = torch.randn(2, requires_grad=False)

def forward(x):
    return torch.relu(x @ W.T + b)

# initialize guess
x = torch.randn(1, 2, requires_grad=True)
optimizer = torch.optim.Adam([x], lr=0.1)

for _ in range(200):
    optimizer.zero_grad()
    loss = ((forward(x) - y_target)**2).mean()
    loss.backward()
    optimizer.step()

print(x)
```
